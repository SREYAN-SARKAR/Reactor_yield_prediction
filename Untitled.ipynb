```python
# ============================================================
# REACTOR YIELD PREDICTION - OPTIMIZED PIPELINE
# ============================================================

import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import (
    KFold,
    RepeatedKFold,
    cross_val_score,
    RandomizedSearchCV
)

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import Ridge
from sklearn.svm import SVR

from sklearn.ensemble import (
    RandomForestRegressor,
    ExtraTreesRegressor,
    GradientBoostingRegressor,
    HistGradientBoostingRegressor,
    VotingRegressor,
    StackingRegressor
)

from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import (
    ConstantKernel,
    Matern,
    WhiteKernel
)

RANDOM_STATE = 42
TARGET = "overall_yield"

BASE_FEATURES = [
    "flow_rate_L_min",
    "concentration_mol_L",
    "inlet_temperature_K",
    "length_m",
    "jacket_temperature_K"
]

print("Libraries loaded successfully.")
```

```python
train = pd.read_csv("train_dataset.csv")
test = pd.read_csv("test_dataset.csv")

print("Training shape:", train.shape)
print("Test shape:", test.shape)

print("\nTraining columns:")
print(train.columns.tolist())

print("\nTest columns:")
print(test.columns.tolist())

display(train.head())
```

```python
assert TARGET in train.columns, f"{TARGET} not found in training data"

for col in BASE_FEATURES:
    assert col in train.columns, f"{col} missing from train"
    assert col in test.columns, f"{col} missing from test"

print("Missing values in train:")
display(train.isnull().sum().to_frame("Missing"))

print("\nMissing values in test:")
display(test.isnull().sum().to_frame("Missing"))

print("\nTarget summary:")
display(train[TARGET].describe())
```

```python
EPS = 1e-8

def create_features(df):
    
    d = df.copy()

    flow = d["flow_rate_L_min"].astype(float)
    conc = d["concentration_mol_L"].astype(float)
    inlet_temp = d["inlet_temperature_K"].astype(float)
    length = d["length_m"].astype(float)
    jacket_temp = d["jacket_temperature_K"].astype(float)

    # ========================================================
    # 1. THERMAL FEATURES
    # ========================================================

    d["temp_diff"] = jacket_temp - inlet_temp

    d["temp_ratio"] = (
        jacket_temp / (inlet_temp + EPS)
    )

    d["relative_temp_diff"] = (
        d["temp_diff"] / (inlet_temp + EPS)
    )

    d["inv_inlet_temp"] = (
        1.0 / (inlet_temp + EPS)
    )

    d["inv_jacket_temp"] = (
        1.0 / (jacket_temp + EPS)
    )

    # ========================================================
    # 2. RESIDENCE-TIME FEATURES
    # ========================================================

    d["residence_proxy"] = (
        length / (flow + EPS)
    )

    d["inv_flow_rate"] = (
        1.0 / (flow + EPS)
    )

    d["flow_length_ratio"] = (
        flow / (length + EPS)
    )

    d["log_residence_proxy"] = (
        np.log1p(d["residence_proxy"])
    )

    # ========================================================
    # 3. PHYSICS-INSPIRED INTERACTIONS
    # ========================================================

    d["thermal_residence"] = (
        d["temp_diff"] * d["residence_proxy"]
    )

    d["concentration_residence"] = (
        conc * d["residence_proxy"]
    )

    d["conc_temp_interaction"] = (
        conc * inlet_temp
    )

    d["conc_jacket_interaction"] = (
        conc * jacket_temp
    )

    d["jacket_flow_interaction"] = (
        jacket_temp * flow
    )

    d["length_flow_interaction"] = (
        length * flow
    )

    # ========================================================
    # 4. CAREFULLY SELECTED NONLINEAR FEATURES
    # ========================================================

    d["flow_squared"] = flow ** 2

    d["conc_squared"] = conc ** 2

    d["temp_diff_squared"] = (
        d["temp_diff"] ** 2
    )

    # Remove numerical infinities if any
    d = d.replace(
        [np.inf, -np.inf],
        np.nan
    )

    return d


print("Feature engineering function created.")
```

```python
train_fe = create_features(train)
test_fe = create_features(test)

print("Original train shape:", train.shape)
print("Feature-engineered train shape:", train_fe.shape)

print("\nNew features:")
print([
    col for col in train_fe.columns
    if col not in train.columns
])

display(train_fe.head())
```

```python
feature_sets = {

    "Baseline": BASE_FEATURES,

    "Temperature": BASE_FEATURES + [
        "temp_diff",
        "temp_ratio",
        "relative_temp_diff"
    ],

    "Residence": BASE_FEATURES + [
        "residence_proxy",
        "inv_flow_rate",
        "flow_length_ratio",
        "log_residence_proxy"
    ],

    "Physics_Core": BASE_FEATURES + [
        "temp_diff",
        "temp_ratio",
        "residence_proxy",
        "thermal_residence",
        "concentration_residence"
    ],

    "Physics_Interactions": BASE_FEATURES + [
        "temp_diff",
        "temp_ratio",
        "residence_proxy",
        "thermal_residence",
        "concentration_residence",
        "conc_temp_interaction",
        "conc_jacket_interaction",
        "jacket_flow_interaction"
    ],

    "Nonlinear": BASE_FEATURES + [
        "temp_diff",
        "residence_proxy",
        "thermal_residence",
        "flow_squared",
        "conc_squared",
        "temp_diff_squared"
    ]
}

for name, cols in feature_sets.items():
    print(
        f"{name}: {len(cols)} features"
    )
```

```python
cv_repeated = RepeatedKFold(
    n_splits=5,
    n_repeats=3,
    random_state=RANDOM_STATE
)

cv_5 = KFold(
    n_splits=5,
    shuffle=True,
    random_state=RANDOM_STATE
)

cv_10 = KFold(
    n_splits=10,
    shuffle=True,
    random_state=RANDOM_STATE
)

print("Cross-validation strategies ready.")
```

```python
feature_results = []

y_full = train[TARGET].values

for feature_set_name, cols in feature_sets.items():

    X_current = train_fe[cols].values

    model = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),

        (
            "model",
            ExtraTreesRegressor(
                n_estimators=500,
                max_features=1.0,
                min_samples_leaf=1,
                random_state=RANDOM_STATE,
                n_jobs=-1
            )
        )
    ])

    scores = -cross_val_score(
        model,
        X_current,
        y_full,
        cv=cv_repeated,
        scoring="neg_root_mean_squared_error",
        n_jobs=-1
    )

    feature_results.append({
        "Feature Set": feature_set_name,
        "Number of Features": len(cols),
        "CV RMSE Mean": scores.mean(),
        "CV RMSE Std": scores.std()
    })


feature_leaderboard = (
    pd.DataFrame(feature_results)
    .sort_values(
        "CV RMSE Mean"
    )
    .reset_index(drop=True)
)

display(feature_leaderboard)
```

```python
best_feature_set_name = (
    feature_leaderboard
    .iloc[0]["Feature Set"]
)

best_feature_cols = (
    feature_sets[best_feature_set_name]
)

print("=" * 60)
print("BEST FEATURE SET")
print("=" * 60)

print(
    "Feature Set:",
    best_feature_set_name
)

print(
    "Number of Features:",
    len(best_feature_cols)
)

print("\nSelected Features:")

for feature in best_feature_cols:
    print("-", feature)
```

```python
X = train_fe[
    best_feature_cols
].values

y = train_fe[
    TARGET
].values

X_test = test_fe[
    best_feature_cols
].values

print("X shape:", X.shape)
print("y shape:", y.shape)
print("X_test shape:", X_test.shape)
```

```python
def cv_rmse(model, cv=cv_repeated, X_data=None, y_data=None):

    if X_data is None:
        X_data = X

    if y_data is None:
        y_data = y

    scores = cross_val_score(
        model,
        X_data,
        y_data,
        cv=cv,
        scoring="neg_root_mean_squared_error",
        n_jobs=-1
    )

    return (
        -scores.mean(),
        scores.std()
    )


def make_pipe(
    estimator,
    scale=False
):

    steps = [
        (
            "imputer",
            SimpleImputer(
                strategy="median"
            )
        )
    ]

    if scale:
        steps.append(
            (
                "scale",
                StandardScaler()
            )
        )

    steps.append(
        (
            "model",
            estimator
        )
    )

    return Pipeline(steps)


print("Helper functions ready.")
```

```python
gpr_kernel = (

    ConstantKernel(
        1.0,
        (1e-3, 1e3)
    )

    *

    Matern(
        length_scale=1.0,
        length_scale_bounds=(
            1e-2,
            1e3
        ),
        nu=1.5
    )

    +

    WhiteKernel(
        noise_level=1.0,
        noise_level_bounds=(
            1e-8,
            1e2
        )
    )
)


candidates = {

    "Ridge (alpha=1)": make_pipe(
        Ridge(alpha=1.0),
        scale=True
    ),

    "Ridge (alpha=10)": make_pipe(
        Ridge(alpha=10.0),
        scale=True
    ),

    "SVR (RBF)": make_pipe(
        SVR(
            kernel="rbf",
            C=100,
            epsilon=1.0,
            gamma="scale"
        ),
        scale=True
    ),

    "RandomForest": make_pipe(
        RandomForestRegressor(
            n_estimators=500,
            random_state=RANDOM_STATE,
            n_jobs=-1
        )
    ),

    "RandomForest (Regularized)": make_pipe(
        RandomForestRegressor(
            n_estimators=500,
            max_depth=6,
            min_samples_leaf=3,
            max_features=0.5,
            random_state=RANDOM_STATE,
            n_jobs=-1
        )
    ),

    "ExtraTrees": make_pipe(
        ExtraTreesRegressor(
            n_estimators=500,
            random_state=RANDOM_STATE,
            n_jobs=-1
        )
    ),

    "GradientBoosting": make_pipe(
        GradientBoostingRegressor(
            n_estimators=300,
            learning_rate=0.05,
            max_depth=3,
            subsample=0.9,
            random_state=RANDOM_STATE
        )
    ),

    "HistGradientBoosting": make_pipe(
        HistGradientBoostingRegressor(
            max_iter=400,
            learning_rate=0.08,
            max_depth=4,
            random_state=RANDOM_STATE
        )
    ),

    "Gaussian Process": make_pipe(
        GaussianProcessRegressor(
            kernel=gpr_kernel,
            normalize_y=True,
            n_restarts_optimizer=3,
            random_state=RANDOM_STATE
        ),
        scale=True
    )
}

print(
    "Number of candidate models:",
    len(candidates)
)
```

```python
rows = []

for name, model in candidates.items():

    mean_rmse, std_rmse = cv_rmse(model)

    rows.append({
        "Model": name,
        "CV RMSE Mean": mean_rmse,
        "CV RMSE Std": std_rmse
    })


comparison = (
    pd.DataFrame(rows)
    .sort_values(
        "CV RMSE Mean"
    )
    .reset_index(drop=True)
)

display(comparison)
```

```python
search_spaces = {

    "GradientBoosting": (

        make_pipe(
            GradientBoostingRegressor(
                random_state=RANDOM_STATE
            )
        ),

        {
            "model__n_estimators": [
                200,
                300,
                500,
                800
            ],

            "model__learning_rate": [
                0.02,
                0.03,
                0.05,
                0.08
            ],

            "model__max_depth": [
                2,
                3,
                4
            ],

            "model__subsample": [
                0.7,
                0.8,
                0.9,
                1.0
            ],

            "model__min_samples_leaf": [
                1,
                2,
                3
            ]
        }
    ),

    "ExtraTrees": (

        make_pipe(
            ExtraTreesRegressor(
                random_state=RANDOM_STATE,
                n_jobs=-1
            )
        ),

        {
            "model__n_estimators": [
                300,
                500,
                800
            ],

            "model__max_depth": [
                None,
                8,
                12,
                20
            ],

            "model__max_features": [
                0.5,
                0.6,
                0.8,
                1.0
            ],

            "model__min_samples_leaf": [
                1,
                2,
                3
            ]
        }
    ),

    "RandomForest": (

        make_pipe(
            RandomForestRegressor(
                random_state=RANDOM_STATE,
                n_jobs=-1
            )
        ),

        {
            "model__n_estimators": [
                300,
                500,
                800
            ],

            "model__max_depth": [
                None,
                6,
                8,
                12
            ],

            "model__max_features": [
                0.4,
                0.5,
                0.6,
                0.8
            ],

            "model__min_samples_leaf": [
                1,
                2,
                3
            ]
        }
    )
}


tuned = {}

for name, (
    model_pipeline,
    parameter_space
) in search_spaces.items():

    search = RandomizedSearchCV(

        estimator=model_pipeline,

        param_distributions=parameter_space,

        n_iter=25,

        cv=cv_repeated,

        scoring="neg_root_mean_squared_error",

        random_state=RANDOM_STATE,

        n_jobs=-1
    )

    search.fit(X, y)

    tuned[
        f"{name} (tuned)"
    ] = search.best_estimator_

    print("\n" + "=" * 60)

    print(
        f"{name} Tuning Complete"
    )

    print(
        f"Best CV RMSE: {-search.best_score_:.4f}"
    )

    print(
        "Best Parameters:"
    )

    print(
        search.best_params_
    )
```

```python
pool = {
    **candidates,
    **tuned
}

individual_scores = {}

for name, model in pool.items():

    individual_scores[name] = (
        cv_rmse(model)
    )


individual_leaderboard = (

    pd.DataFrame([

        {
            "Model": name,
            "CV RMSE Mean": result[0],
            "CV RMSE Std": result[1]
        }

        for name, result
        in individual_scores.items()

    ])

    .sort_values(
        "CV RMSE Mean"
    )

    .reset_index(
        drop=True
    )
)

display(
    individual_leaderboard
)
```

```python
top_3_names = (
    individual_leaderboard[
        "Model"
    ]
    .head(3)
    .tolist()
)

top_3_models = [

    (
        name.replace(
            " ",
            "_"
        ),

        pool[name]
    )

    for name in top_3_names
]

print("=" * 60)

print(
    "TOP 3 INDIVIDUAL MODELS"
)

print("=" * 60)

for i, name in enumerate(
    top_3_names,
    start=1
):
    print(
        f"{i}. {name}"
    )
```

```python
weight_sets = [

    (1, 1, 1),

    (2, 1, 1),

    (1, 2, 1),

    (1, 1, 2),

    (2, 2, 1),

    (2, 1, 2),

    (1, 2, 2),

    (3, 2, 1),

    (3, 1, 2),

    (4, 2, 1),

    (5, 3, 2)
]


ens = {

    "Voting (top 2)": VotingRegressor(
        estimators=top_3_models[:2]
    ),

    "Voting (top 3 equal)": VotingRegressor(
        estimators=top_3_models
    ),

    "Stacking (Ridge meta)": StackingRegressor(

        estimators=top_3_models,

        final_estimator=Ridge(
            alpha=1.0
        ),

        cv=cv_5,

        n_jobs=-1
    )
}


for weights in weight_sets:

    ensemble_name = (
        f"Voting Weights {weights}"
    )

    ens[
        ensemble_name
    ] = VotingRegressor(

        estimators=top_3_models,

        weights=weights
    )


print(
    "Number of ensemble models:",
    len(ens)
)
```

```python
ensemble_results = []

for name, model in ens.items():

    mean_rmse, std_rmse = (
        cv_rmse(model)
    )

    ensemble_results.append({

        "Model": name,

        "CV RMSE Mean": mean_rmse,

        "CV RMSE Std": std_rmse
    })


ensemble_leaderboard = (

    pd.DataFrame(
        ensemble_results
    )

    .sort_values(
        "CV RMSE Mean"
    )

    .reset_index(
        drop=True
    )
)

display(
    ensemble_leaderboard
)
```

```python
all_models = {
    **pool,
    **ens
}


scores = {}

for name, model in all_models.items():

    scores[name] = (
        cv_rmse(model)
    )


leaderboard = (

    pd.DataFrame([

        {
            "Model": name,

            "CV RMSE Mean": result[0],

            "CV RMSE Std": result[1]
        }

        for name, result
        in scores.items()

    ])

    .sort_values(
        "CV RMSE Mean"
    )

    .reset_index(
        drop=True
    )
)


display(
    leaderboard
)


best_name = (
    leaderboard
    .iloc[0]["Model"]
)

best_model = (
    all_models[
        best_name
    ]
)

best_mean, best_std = (
    scores[
        best_name
    ]
)


print("\n" + "=" * 60)

print(
    "FINAL MODEL SELECTION"
)

print("=" * 60)

print(
    "Selected Model:",
    best_name
)

print(
    f"Best CV RMSE: {best_mean:.4f}"
)

print(
    f"CV RMSE Std: {best_std:.4f}"
)
```

```python
best_individual_name = (
    individual_leaderboard
    .iloc[0]["Model"]
)

best_ensemble_name = (
    ensemble_leaderboard
    .iloc[0]["Model"]
)


final_candidates = {

    "Best Individual": pool[
        best_individual_name
    ],

    "Best Ensemble": ens[
        best_ensemble_name
    ],

    "Overall Winner": best_model
}


seeds = [
    42,
    123,
    456,
    789,
    2026
]


stability_results = []


for model_name, model in final_candidates.items():

    for seed in seeds:

        cv_test = RepeatedKFold(

            n_splits=5,

            n_repeats=3,

            random_state=seed
        )


        scores_seed = (
            -cross_val_score(

                model,

                X,

                y,

                cv=cv_test,

                scoring=(
                    "neg_root_mean_squared_error"
                ),

                n_jobs=-1
            )
        )


        stability_results.append({

            "Model": model_name,

            "Seed": seed,

            "RMSE Mean": (
                scores_seed.mean()
            ),

            "RMSE Std": (
                scores_seed.std()
            )
        })


stability_df = pd.DataFrame(
    stability_results
)


stability_summary = (

    stability_df

    .groupby(
        "Model"
    )

    .agg(

        Average_RMSE=(
            "RMSE Mean",
            "mean"
        ),

        Stability=(
            "RMSE Mean",
            "std"
        ),

        Best_RMSE=(
            "RMSE Mean",
            "min"
        ),

        Worst_RMSE=(
            "RMSE Mean",
            "max"
        )
    )

    .sort_values(
        "Average_RMSE"
    )
)


display(
    stability_summary
)
```

```python
print("=" * 70)

print(
    "FINAL OPTIMIZED MODEL SUMMARY"
)

print("=" * 70)

print(
    f"Best Feature Set : "
    f"{best_feature_set_name}"
)

print(
    f"Number of Features : "
    f"{len(best_feature_cols)}"
)

print(
    f"Selected Model : "
    f"{best_name}"
)

print(
    f"Best CV RMSE : "
    f"{best_mean:.4f}"
)

print(
    f"CV RMSE Std : "
    f"{best_std:.4f}"
)

print("=" * 70)
```

```python
best_model.fit(
    X,
    y
)

print(
    "Final model successfully trained."
)

print(
    "Training rows:",
    X.shape[0]
)

print(
    "Number of features:",
    X.shape[1]
)
```
